```python
import os
import geopandas as gpd
```

```python
%%time
print(f"Inicia la lectura del mapa")
ColombiaManzanas =  gpd.read_file('ShapeManzanas/MGN_ANM_MANZANA.shp')
print(f"Finaliza la lectura del mapa")
```

## Creación de los shapes por código de municipio
El siguiente código crea un shape por código de municipio y los almacena en la carpeta Manzana

```python
# Comentar la siguiente linea si la carpeta ya está creada
os.mkdir("Manzana")
for CodigoMunicipio in ColombiaManzanas.MPIO_CDPMP.unique():
    print(f"Creando municipio {CodigoMunicipio}")
    subdf = ColombiaManzanas[ColombiaManzanas["MPIO_CDPMP"] == CodigoMunicipio]
    subdf.to_file(f'Manzana/{CodigoMunicipio}.shp')

print("Termina el procesamiento")
```
